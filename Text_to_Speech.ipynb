```python
from tkinter import *
from gtts import gTTS
from playsound3 import playsound
```

```python
root = Tk()
root.geometry("350x300")
root.configure(bg='ghost white')
root.title("TEXT TO SPEECH")
```

```python
Label(root, text = "TEXT_TO_SPEECH", font = "arial 20 bold", bg='white smoke').pack()
Label(text ="", font = 'arial 15 bold', bg ='white smoke' , width = '20').pack(side = 'bottom')

Msg = StringVar()
Label(root,text ="Enter Text", font = 'arial 15 bold', bg ='white smoke').place(x=20,y=60)

entry_field = Entry(root, textvariable = Msg ,width ='50')
entry_field.place(x=20,y=100)
```

```python
def Text_to_speech():
    Message = entry_field.get()
    speech = gTTS(text = Message)
    speech.save('audio.mp3')
    playsound('audio.mp3')
```

```python
def Exit():
    root.destroy()
```

```python
def Reset():
    Msg.set("")
```

```python
Button(
    root,
    text="PLAY",
    font="Arial 15 bold",
    command=Text_to_speech,
    width=4
).place(x=25, y=140)

Button(
    root,
    text="EXIT",
    font="Arial 15 bold",
    command=Exit,
    width=4,
    bg="OrangeRed1"
).place(x=100, y=140)

Button(
    root,
    text="RESET",
    font="Arial 15 bold",
    command=Reset,
    width=6
).place(x=175, y=140)

root.mainloop()
```
